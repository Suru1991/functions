Ques.1)What is the difference between a function and a method in Python?

Answer-Functions can be called only by its name, as it is defined independently. But methods can't be called by its name only, we need to invoke the class by a reference of that class in which it is defined, i.e. method is defined within a class and hence they are dependent on that class.

In short , function is a free, standalone block of code and method is function tied to an object.A method is defined inside a class.

```python
#example of difference in function and method
# function
def add(a, b):
    return a + b

print(add(3, 5))   # 8
```

```python
#method
class Calculator:
    def add(self, a, b):   # 'self' refers to the object
        return a + b

# create object
calc = Calculator()

print(calc.add(3, 5))   # 8
#Here add is a method of the Calculator class, while in the previous case add was just a function.
```

Ques.5)What are iterators in Python and how do they differ from iterables?

Answer-An iterable is any Python object capable of returning its elements one at a time.These objects implement the __iter__() method.Examples: list, tuple, string, dict, set.

```python
#showing iterable

my_list = [1, 2, 3]

for num in my_list:   # my_list is an iterable
    print(num)
```

An iterator is the object that actually produces the values from an iterable, one at a time.Created when we call iter() on an iterable.Implements two methods:

*   __iter__() → returns the iterator object itself
*   __next__() → returns the next value, raises StopIteration when done



```python
#showing iterator
my_list = [1, 2, 3]
it = iter(my_list)   # create iterator

print(next(it))  # 1
print(next(it))  # 2
print(next(it))  # 3
# next(it) -> raises StopIteration
```

```python
#Example-Iterable vs Iterator
# List is an iterable
nums = [10, 20, 30]

# Convert iterable into iterator
nums_iter = iter(nums)

print(next(nums_iter))
print(next(nums_iter))
print(next(nums_iter))
```

Ques.6) Explain the concept of generators in Python and how they are defined.

Answer-A **generator** is a special type of iterator.Instead of storing all values in memory (like lists), it yields values one at a time when needed.This makes them memory-efficient for working with large data.

**How generators are defined?**


*   Using a function with yield


    *   If a function contains yield instead of return, it becomes a generator.
    *   Each time yield is called, the function pauses and saves its state.


    *   When called again, it resumes from where it left off.



*   Using Generator expressions

    *   Similar to list comprehensions, but use () instead of [].


    *   Produces values lazily (on demand).









```python
# example-generator defined using yield
def count_up_to(n):
    i = 1
    while i <= n:
        yield i
        i += 1

# Using the generator
for num in count_up_to(5):
    print(num)
```

```python
#example-generator defined using  generator expression

squares = (x*x for x in range(5))
for val in squares:
    print(val)
```

Ques.7)What are the advantages of using generators over regular functions?

Answer-Generators give you several advantages over regular functions.

** Memory Efficiency**

      *   Regular functions with return often create and return entire data structures.
      *   Generators yield one value at a time, so they don't need to store the whole sequence in memory.
      *   Useful when working with large datasets.





```python
#showing memory efficiency in generator with example in comparison to function
# Regular function (stores whole list in memory)
def get_numbers(n):
    return [i for i in range(n)]

# Generator (yields one value at a time)
def generate_numbers(n):
    for i in range(n):
        yield i
```


  

*   **Lazy Evaluation**

    *   Generators produce values on demand, instead of computing everything upfront.
    *   Makes your program faster when you don’t need all results immediately.


*   **Infinite Sequences**



    *   Regular functions can’t handle infinite data because they must return all values at once.
    *   Generators can model infinite streams (like sensor data or logs).





```python
def infinite_numbers():
    i = 1
    while True:
        yield i
        i += 1

gen = infinite_numbers()
print(next(gen))  # 1
print(next(gen))  # 2
```



*   **Pipeline Processing**


    *   Generators can be chained together like pipelines.





```python
#showing pipeline processing in generators
def numbers(n):
    for i in range(n):
        yield i

def square(seq):
    for num in seq:
        yield num * num

print(list(square(numbers(5))))  # [0, 1, 4, 9, 16]
```

Ques.9) Explain the purpose and usage of the `map()` function in Python.

Answer-The map() function is a very useful built-in Python function for applying operations to all elements in an iterable.


**Purpose of map()**



*   map() applies a given function to each item of an iterable (like a list, tuple, etc.).

*   Returns a map object, which is an iterator.
*   Can be converted into a list, tuple, or set if needed.


**Usage of map()**--map() is used to apply a function to every element of an iterable (or multiple iterables) in a clean, efficient way.

1)**Using map with one iterable**


```python
def square(x):
    return x * x

nums = [1, 2, 3, 4]
squared_nums = map(square, nums)

print(list(squared_nums))  # [1, 4, 9, 16]
```

2)**Using map with a lambda**

```python
nums = [5, 6, 7, 8]
squared_nums = list(map(lambda x: x**2, nums))
print(squared_nums)  # [25, 36, 49, 64]
```

3)**Using map with multiple iterables**

```python
a = [1, 2, 3]  # here i have taken two iterables i.e.,lists
b = [4, 5, 6]

sum_ab = list(map(lambda x, y: x + y, a, b))
print(sum_ab)  # [5, 7, 9]
```

Ques.10) What is the difference between `map()`, `reduce()`, and `filter()` functions in Python?

Answer-These three are functional programming tools in Python and are often used with iterables.

**map()**



*  The purpose of map()-Applies a function to each item.

*   map() returns an iterator.


*   key idea-It transforms the iterable



```python
nums = [1, 2, 3, 4]
squared = list(map(lambda x: x**2, nums))
print(squared)  # [1, 4, 9, 16]
```

**filter()**



*  Purpose of filter()-Filter out elements based on a condition (function returns True/False).

*   Returns a filter object (an iterator).
*   Key Idea: Selects elements from iterable.






```python
nums = [1, 2, 3, 4, 5, 6]
evens = list(filter(lambda x: x % 2 == 0, nums))
print(evens)  # [2, 4, 6]
```

**reduce()**



*   Imported from functools module

*   Reduce an iterable to a single cumulative value by applying a binary function.
*   It will return a single value


*   Key Idea:combines elements step bystep



```python
from functools import reduce

nums = [1, 2, 3, 4]
sum_total = reduce(lambda x, y: x + y, nums)
print(sum_total)  # 10
```

```python
#Ques.4)Write a Python function that checks if a given number is prime or not from 1 to 200.

def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

# --- Take input from user ---
num = int(input("Enter a number between 1 and 200: "))

if 1 <= num <= 200:
    if is_prime(num):
        print(num, "is a prime number.")
    else:
        print(num, "is not a prime number.")
else:
    print("Please enter a number in the range 1 to 200.")
```

```python
#Ques.5) Create an iterator class in Python that generates the Fibonacci sequence up to a specified number of terms.

class FibonacciIterator:
    def __init__(self, n_terms):
        self.n_terms = n_terms  # total number of terms to generate
        self.index = 0          # current position in the sequence
        self.a = 0              # first Fibonacci number
        self.b = 1              # second Fibonacci number

    def __iter__(self):
        return self

    def __next__(self):
        if self.index >= self.n_terms:
            raise StopIteration  # stop iteration when we reach n_terms
        if self.index == 0:
            self.index += 1
            return self.a
        elif self.index == 1:
            self.index += 1
            return self.b
        else:
            self.index += 1
            fib = self.a + self.b
            self.a, self.b = self.b, fib
            return fib

# Example usage
n = 10
fib_iter = FibonacciIterator(n)
for num in fib_iter:
    print(num, end=" ")



```

```python
#Ques.6)Write a generator function in Python that yields the powers of 2 up to a given exponent.

def powers_of_two(max_exponent):
    """
    Generator that yields powers of 2 from 2^0 up to 2^max_exponent.
    """
    for exponent in range(max_exponent + 1):
        yield 2 ** exponent

# Example usage:
max_exp = 5
for value in powers_of_two(max_exp):
    print(value, end=" ")
```

```python
#Ques.Implement a generator function that reads a file line by line and yields each line as a string.

def read_file_line_by_line(filename):
    """
    Generator that reads a file line by line and yields each line.
    """
    try:
        with open(filename, 'r') as f:
            for line in f:
                yield line.strip()  # Yield each line, stripped of leading/trailing whitespace
    except FileNotFoundError:
        print(f"Error: File '{filename}' not found.")
    except Exception as e:
        print(f"An error occurred: {e}")

# Example usage:
# Create a dummy file for demonstration
with open("my_file.txt", "w") as f:
    f.write("This is the first line.\n")
    f.write("This is the second line.\n")
    f.write("And this is the third line.")

# Use the generator to read the file
for line in read_file_line_by_line("my_file.txt"):
    print(line)
```
